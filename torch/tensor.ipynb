GPU tensors

```python
torch.cuda.is_available()
```

```python
if torch.cuda.is_available():
    print("GPU available, but we will attempt to use TPU")
else:
    print("GPU not available")
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")
```

```python
torch.rand((2,2), device = device)
```

```python
import time

size = 10000

mc1 = torch.randn(size, size)
mc2 = torch.randn(size, size)

starttime = time.time()

mc3 = torch.matmul(mc1, mc2)

endtime = time.time()

print(f"Time on CPU = {endtime - starttime}")

mg1 = mc1.to(device)
mg2 = mc2.to(device)

starttime = time.time()

mg3 = torch.matmul(mg1, mg2)

endtime = time.time()

print(f"Time on GPU = {endtime - starttime}")
```
