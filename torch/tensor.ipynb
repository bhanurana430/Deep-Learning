#Perceptron Example for AUTOGRAD

```python
x = torch.tensor(6.7) #input tensor
y = torch.tensor(0.0) #True label

w = torch.tensor(1.0)
b = torch.tensor(0.0)
```

```python
#binary cross entropy for scalar
def bce(pred, target):
    epsilion = 1e-8
    pred = torch.clamp(pred, epsilion, 1-epsilion)
    return -(target*torch.log(pred) + (1-target)*torch.log(1-pred))
```

```python
#forward pass
z = w*x + b
y_pred = torch.sigmoid(z)

loss = bce(y_pred, y)
```

```python
loss
```

```python
#Derivatives/ Backprop
#Manual - skip

# Auto grad
x = torch.tensor(6.7) #input tensor
y = torch.tensor(0.0) #True label

#We didnot put requiresgrad true bcz we dont need it with respect to x or y, we need wrt w,b

w = torch.tensor(1.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)
```

```python
w
```

```python
b
```

```python
#forward
z = w*x + b
z
```

```python
y_pred = torch.sigmoid(z)
y_pred
```

```python
loss = bce(y_pred, y)
loss
```

```python
loss.backward()
```

```python
w.grad
```

```python
b.grad
```
